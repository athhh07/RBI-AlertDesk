```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('data/raw/rbi_circular_texts.csv',low_memory=False)
df.shape
```

```python
urgent_words=[
    "shall",
    "must",
    "mandatory",
    "penalty",
    "within",
    "immediately",
    "provide",
    "as soon as possible",
    "early",
    "waiting",
    "urgent",
    "strictly"
]

df["full_text"]=df["full_text"].fillna("").astype(str)

df["word_count"]=df["full_text"].str.split().str.len()

df["urgency_score"]=df["full_text"].apply(
    lambda text: sum(
        text.lower().count(word)
        for word in urgent_words
    )
)

df["urgency_density"]=(
    df["urgency_score"]
    /
    df["word_count"]
)*1000

df["urgency_density"]=df["urgency_density"].fillna(0)
```

```python
df[[
    "word_count",
    "urgency_score",
    "urgency_density"
]].describe()
```

```python
q20=df["urgency_density"].quantile(0.20)
q80=df["urgency_density"].quantile(0.80)

print("20th Percentile:",q20)
print("80th Percentile:",q80)
```

```python
high_pool=df[
    df["urgency_density"]>=q80
]

medium_pool=df[
    (df["urgency_density"]>q20)
    &
    (df["urgency_density"]<q80)
]

low_pool=df[
    df["urgency_density"]<=q20
]

print("High:",len(high_pool))
print("Medium:",len(medium_pool))
print("Low:",len(low_pool))
```

```python
high=high_pool.sample(
    n=60,
    random_state=42
)

print(high.shape)
```

```python
medium=medium_pool.sample(
    n=120,
    random_state=42
)

print(medium.shape)
```

```python
low=low_pool.sample(
    n=40,
    random_state=42
)

print(low.shape)
```

```python
largest=df.nlargest(
    15,
    "word_count"
)

largest[[
    "title",
    "word_count"
]].head()
```

```python
smallest=df.nsmallest(
    15,
    "word_count"
)

smallest[[
    "title",
    "word_count"
]].head()
```

```python
labeling_df=pd.concat([
    high,
    medium,
    low,
    largest,
    smallest
])

labeling_df=labeling_df.drop_duplicates()

labeling_df=labeling_df.reset_index(
    drop=True
)

print(labeling_df.shape)
```

```python
labeling_df["label"]=""
labeling_df["notes"]=""
```

```python
labeling_df.to_excel(
    "data/manual_labeling_sample.xlsx",
    index=False
)

print("File Saved")
```
