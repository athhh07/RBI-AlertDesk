### Downloading 5 sample pdfs:

```python
import pandas as pd
import requests
import os

df = pd.read_csv("data/raw/rbi_circulars_combined.csv")

os.makedirs("pdfs", exist_ok=True)

for i in range(5):

    pdf_url = df.iloc[i]["pdf_url"]

    response = requests.get(pdf_url, timeout=30)

    with open(f"pdfs/{i}.pdf", "wb") as f:
        f.write(response.content)

print("Done")
```

### Extracting text from one pdf:

```python
import fitz
doc = fitz.open("pdfs/1.pdf")
text = ""
for page in doc:
    text += page.get_text()
print(text[:3000])
```

```python
import pandas as pd
import fitz
import requests
import os
from tqdm import tqdm

df = pd.read_csv("data/raw/rbi_circulars_combined.csv")

records = []

os.makedirs("pdfs", exist_ok=True)

for idx, row in tqdm(df.iterrows(), total=len(df)):

    pdf_path = f"pdfs/{idx}.pdf"

    try:

        r = requests.get(row["pdf_url"], timeout=30)

        with open(pdf_path, "wb") as f:
            f.write(r.content)

        doc = fitz.open(pdf_path)

        text = ""

        for page in doc:
            text += page.get_text()

        records.append({
            "date": row["date"],
            "circular_id": row["circular_id"],
            "title": row["title"],
            "full_text": text
        })

    except Exception as e:

        print(idx, e)

output = pd.DataFrame(records)
output.to_csv("data/circular_texts.csv",index=False)

print(output.shape)
```

```python
output["text_length"] = output["full_text"].str.len()
print(output["text_length"].describe())
```
