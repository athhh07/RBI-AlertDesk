## Section 4- Text Cleaning:

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords

stop_words=set(stopwords.words("english"))
def clean_text(text):
    text=str(text).lower()
    text=re.sub(r'[^a-zA-Z ]',' ',text)
    words=text.split()
    words=[w for w in words if w not in stop_words]
    return " ".join(words)
df["clean_text"]=df["full_text"].apply(clean_text)
```

## Section 5- RBI Domain Analysis:

```python
keywords=[
    "shall",
    "must",
    "mandatory",
    "penalty",
    "within",
    "immediately",
    "compliance",
    "fraud",
    "kyc",
    "customer",
    "nbfc",
    "bank"
]
```

```python
keyword_freq={}

for word in keywords:
    keyword_freq[word]=df["clean_text"].str.count(word).sum()

keyword_freq=pd.Series(keyword_freq)

keyword_freq.sort_values(ascending=False)
```

### Key Findings:

#### - Regulatory language is highly directive, with the term "shall" appearing over 58,000 times.
#### - Customer-related topics (customer, KYC) are among the most frequent themes, indicating a strong focus on customer protection and due diligence.
#### - NBFCs are frequently referenced, suggesting that many circulars apply beyond traditional banks.
#### - Urgency indicators such as "mandatory", "immediately", and "penalty" occur much less frequently, which may help distinguish high-priority circulars during classification.

```python
keyword_freq=keyword_freq.sort_values(
    ascending=False
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=keyword_freq.index,
    y=keyword_freq.values
)

plt.xticks(rotation=45)
plt.title("Compliance Related Keywords")
plt.show()
```

### Keywords presense analysis:

```python
keyword_presence={}

for word in keywords:
    keyword_presence[word]=(
        df["clean_text"]
        .str.contains(word,case=False)
        .sum()
    )

keyword_presence=pd.Series(keyword_presence)
keyword_presence.sort_values(
    ascending=False
)
```

```python
keyword_presence=keyword_presence.sort_values(
    ascending=False
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=keyword_presence.index,
    y=keyword_presence.values
)

plt.xticks(rotation=45)
plt.title("Circulars Mentioning Each Keyword")
plt.show()
```

## Section 6- Prepare for Labeling:

```python
urgent_words=[
    "shall",
    "must",
    "mandatory",
    "penalty",
    "within",
    "immediately",
    "urgent",
    "strictly",
    "compliance",
    "required"
]
```

### Raw urgency score:

```python
for word in urgent_words:
    df[word+"_count"]=df["clean_text"].str.count(word)

df["urgency_score"]=df[
    [word+"_count" for word in urgent_words]
].sum(axis=1)

df[["title","urgency_score"]].head()
```

### Urgency density:

```python
df["urgency_density"]=(
    df["urgency_score"]/
    df["word_count"]
)*1000

df[[
    "title",
    "word_count",
    "urgency_score",
    "urgency_density"
]].head()
```

### Urgency density stats:

```python
df["urgency_density"].describe(
    percentiles=[0.90,0.95,0.99]
)
```

### Top 30 urgent candidates:

```python
df.nlargest(
    30,
    "urgency_density"
)[[
    "date",
    "title",
    "word_count",
    "urgency_score",
    "urgency_density"
]]
```

### Urgency density distribution:

```python
plt.figure(figsize=(10,5))

sns.histplot(
    df["urgency_density"],
    bins=50
)
plt.title("Urgency Density Distribution")
plt.show()
```

### Observation:

#### - Raw urgency scores are strongly influenced by document length.
#### - Large master directions contain many occurrences of regulatory terms such as "shall" and "must", leading to inflated scores.
#### - Urgency density normalizes keyword counts by document size and provides a fairer comparison across circulars.
#### - Circulars with high urgency density will be prioritized during manual labeling.
