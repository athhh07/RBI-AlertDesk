## The purpose of this is to compare multiple models using the same train/test split.

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report
from sklearn.metrics import confusion_matrix
from sklearn.metrics import ConfusionMatrixDisplay

import matplotlib.pyplot as plt

import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_excel("data/using/manual_labeling_sample.xlsx")

df=df[[
    "full_text",
    "label"
]]

df=df.dropna()
df["label"]=df["label"].str.strip()
df.head()
```

```python
df.tail()
```

```python
df.shape
print(df["label"].value_counts())
```

```python
X=df["full_text"] 
y=df["label"]
```

```python
X_train,X_test,y_train,y_test=train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)
```

```python
tfidf=TfidfVectorizer(
    max_features=5000,
    stop_words="english",
    ngram_range=(1,2)
)
X_train=tfidf.fit_transform(X_train)
X_test=tfidf.transform(X_test)
```

### Logistic Regression:

```python
lr=LogisticRegression(
    class_weight="balanced",
    max_iter=1000,
    random_state=42
)
lr.fit(X_train,y_train)

lr_pred=lr.predict(X_test)
```

```python
print("LOGISTIC REGRESSION")
print()
print(classification_report(y_test,lr_pred))
```

```python
lr_accuracy=accuracy_score(
    y_test,
    lr_pred
)
print(lr_accuracy)
```

### Linear SVM:

```python
svm=LinearSVC(
    class_weight="balanced",
    random_state=42
)
svm.fit(X_train,y_train)
svm_pred=svm.predict(X_test)
```

```python
print("LINEAR SVM")
print()
print(classification_report(y_test,svm_pred))
```

```python
svm_accuracy=accuracy_score(
    y_test,
    svm_pred
)
print(svm_accuracy)
```

### Random Forest:

```python
rf=RandomForestClassifier(
    n_estimators=300,
    class_weight="balanced",
    random_state=42
)

rf.fit(X_train,y_train)
rf_pred=rf.predict(X_test)
```

```python
print("RANDOM FOREST")
print()
print(classification_report(y_test,rf_pred))
```

```python
rf_accuracy=accuracy_score(
    y_test,
    rf_pred
)
print(rf_accuracy)
```

### Comparing Models:

```python
results=pd.DataFrame({

    "Model":[
        "Logistic Regression",
        "Linear SVM",
        "Random Forest"
    ],
    "Accuracy":[
        lr_accuracy,
        svm_accuracy,
        rf_accuracy
    ]

})
results.sort_values("Accuracy",ascending=False)
```

```python
plt.figure(figsize=(8,5))
plt.bar(
    results["Model"],
    results["Accuracy"]
)

plt.title("Model Comparison")
plt.ylabel("Accuracy")
plt.ylim(0,1)
plt.show()
```

```python
best_model=results.sort_values("Accuracy",ascending=False).iloc[0]
best_model
```

```python
print("Best Model:",best_model["Model"])
print("Accuracy:",round(best_model["Accuracy"],4))
```

```python
results.to_csv("results/model_comparison.csv",index=False)
print("Saved")
```

### My Explaination:
#### Three classifiers (Logistic Regression, Linear SVM, and Random Forest) were evaluated using identical TF-IDF features and a stratified train-test split. Random Forest achieved the highest test accuracy of 87.5%, outperforming Linear SVM (85.4%) and Logistic Regression (83.3%). Based on the  evaluation metrics, Random Forest was selected as the final model.
