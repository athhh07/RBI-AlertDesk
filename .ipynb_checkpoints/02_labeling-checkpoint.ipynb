```python
import pandas as pd
import numpy as np

pd.set_option("display.max_colwidth",500)
```

```python
df=pd.read_csv('data/using/rbi_circulars_eda.csv')
df.columns
```

### Checking urgency distribution:

```python
df["urgency_density"].describe(
    percentiles=[0.20,0.40,0.60,0.80]
)
```

### Creating sampling buckets:

```python
q20=df["urgency_density"].quantile(0.20)
q80=df["urgency_density"].quantile(0.80)
print("20th percentile:",q20)
print("80th percentile:",q80)
```

### High urgency samples:

```python
high=df[
    df["urgency_density"]>=q80
].sample(
    n=60,
    random_state=42
)
print(high.shape)
```

### Medium urgency samples:

```python
medium=df[
    (df["urgency_density"]>q20)
    &
    (df["urgency_density"]<q80)
].sample(
    n=120,
    random_state=42
)
print(medium.shape)
```

### Low urgency samples:

```python
low=df[
    df["urgency_density"]<=q20
].sample(
    n=40,
    random_state=42
)
print(low.shape)
```

### Largest documents (top 15) :

```python
largest=df.nlargest(
    15,
    "word_count"
)
largest[[
    "title",
    "word_count"
]]
```

### Shortest documents (top 15) :

```python
shortest=df.nsmallest(
    15,
    "word_count"
)
shortest[[
    "title",
    "word_count"
]]
```

### Combibing the samples:

```python
labeling_df=pd.concat([
    high,
    medium,
    low,
    largest,
    shortest
])
labeling_df=labeling_df.drop_duplicates(
    subset="circular_id"
)
print(labeling_df.shape)
```

### Shuffling the current review order:

```python
labeling_df=labeling_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
labeling_df.head()
```

### Add label columns:

```python
labeling_df["label"]=""
labeling_df["notes"]=""
labeling_df.head()
```

```python
labeling_df.to_csv(
    "data/using/manual_labeling_review.csv",
    index=False
)
print("Review file created...!")
```

## Label Definitions

INFO
-----
No action required.
Informational updates, publications, announcements.

Examples:
- Exchange rate publication
- Statistical releases
- Clarifications

OPERATIONAL
-----------
Requires process or policy changes.

Examples:
- KYC updates
- Reporting changes
- New compliance framework

URGENT
------
Immediate action required.

Indicators:
- Deadline specified
- Penalty mentioned
- Immediate implementation
- Regulatory enforcement risk
